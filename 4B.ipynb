```python
import numpy as np
import pandas as pd
import re

from gensim.models import Word2Vec
```

```python
data = pd.read_csv("twitter_dataset.csv")

print(data.head())
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "",text)
    text = re.sub(r"[^a-z\s]", " ", text)
    return text.split()


sentences = data["Text"].apply(clean_text)

print(sentences.head())
```

```python
sentences = [words for words in sentences if len(words)>0]
print("Total Tweets:", len(sentences))
```

```python
model = Word2Vec(
    sentences = sentences,
    vector_size = 100,
    window = 5,
    min_count=2,
    sg=0,
    workers=4,
    epochs=10
)
```

```python
words = list(model.wv.index_to_key)
print("Vocabulary Size:", len(words))
print("First 20 Words:", words[:20])
```

```python
word = words[0]
print("Word:", word)
print("Embedding:", model.wv[word])
print("Embedding Shape:", model.wv[word].shape)
```

```python
word = words[3]
similar_words = model.wv.most_similar(word,topn=5)
print("similar words to:", word)

for w, score in similar_words:
    print(w,score)
```

```python
word1 = words[0]
word2 = words[1]

similarity = model.wv.similarity(word1,word2)
print("Word 1:", word1)
print("Word 2:", word2)
print("Similarity:", similarity)
```
