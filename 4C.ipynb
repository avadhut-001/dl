```python
import numpy as np
import pandas as pd
import re
from gensim.models import Word2Vec
```

```python
with open("document3.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)    
```

```python
text = text.lower()

text = re.sub(r"[^a-z\s]", " ", text)

sentences = re.split(r"[.!?]+", text)

sentences = [s.split() for s in sentences if s.strip()]

print(sentences)
```

```python
model = Word2Vec(
    sentences = sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,
    epochs=100
)
```

```python
words = list(model.wv.index_to_key)
print("Vocabulary Size:", len(words))
print("Words:", words)
```

```python
word = words[1]

print(word)
```

```python
word = "influenza"

print("Word:", word)
print("Embedding:", model.wv[word])
print("Embedding Shape:", model.wv[word].shape)
```

```python
word = "virus"

similar_words = model.wv.most_similar(word, topn=5)

print("Similar words to:", word)
for w, score in similar_words:
    print(w, score)
```

```python
word1 = "influenza"
word2 = "virus"

similarity = model.wv.similarity(word1, word2)

print("Word 1:", word1)
print("Word 2:", word2)
print("Similarity:", similarity)
```
